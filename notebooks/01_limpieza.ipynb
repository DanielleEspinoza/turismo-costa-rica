# Limpieza de datos: llegadas internacionales a Costa Rica


Este notebook prepara los datos de llegadas internacionales de turistas para su análisis.

- **Fuente:** Instituto Costarricense de Turismo (ICT), [Informes estadísticos](https://www.ict.go.cr/es/estadisticas/informes-estadisticos.html)
- **Archivo:** `Series_sitio_web_ICT_2025.xlsx`
- **Fecha de descarga:** 30/09/2026
- **Contenido:** llegadas internacionales por año y por país de origen.

**Objetivo:** limpiar y dar formato a los datos para guardarlos en `data/processed/` y usarlos en el análisis exploratorio.

## 1. Carga de datos

```python
import pandas as pd
```

```python
df = pd.read_excel('../data/raw/Series_sitio_web_ICT_2025.xlsx', sheet_name='Todas_vías', header=None)
```

## 2. Exploración inicial 

```python
df.head(15)
```

```python
df.tail(10)
```

### Observaciones

La hoja conserva el formato visual del Excel, por lo que los datos no empiezan en la primera fila:

- **Filas 0 a 10:** títulos y encabezados del cuadro. Los datos empiezan en la fila 11, con el TOTAL.
- **Columna 1:** regiones (TOTAL, AMÉRICA DEL NORTE, etc.), que funcionan como subtotales.
- **Columna 2:** países. Solo tienen valor en las filas de país; en las filas de región aparecen vacías.
- **Columnas 3 a 11:** llegadas de 2017 a 2025.
- **Columnas 12 a 27:** variaciones entre años, absolutas y porcentuales.
- **Últimas filas:** notas al pie y la fuente.
- **Columna 0:** vacía.

## 3. Limpieza

### Selección de filas y columnas

- **Filas 11 a 77:** contienen los datos de llegadas, desde el TOTAL hasta "OTRAS ZONAS". Las filas anteriores son títulos del cuadro y las posteriores son notas al pie y la fuente.
- **Columnas 1 a 11:** región, país y las llegadas de 2017 a 2025. Las columnas restantes son variaciones entre años (absolutas y porcentuales), que se descartan porque pueden calcularse a partir de los datos anuales.

```python
datos = df.iloc[11:78,1:12].copy()
datos.head()
```

```python
datos.shape
```

### Nombres de columnas

Se asignan nombres descriptivos a las columnas: `region`, `pais` y un año por cada columna de llegadas (2017 a 2025).

```python
datos.columns = ['region', 'pais', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']
datos.head()
```

### Región de cada país

En el archivo original, la región solo aparece en la fila de su subtotal, y los países debajo quedan vacíos en esa columna. Con `ffill()` se rellena cada vacío con el último valor de la fila superior, de modo que cada país quede asociado a su región.

```python
datos['region'] = datos['region'].ffill()
datos.head()
```

### Filas que no corresponden a países

- **TOTAL y subtotales de región:** se eliminan porque pueden calcularse sumando los países.
- **Filas separadoras:** se eliminan porque no contienen información.
- **OTRAS ZONAS:**  se conserva. Agrupa Asia y el Pacífico, África y Oriente Medio, con alrededor de 60 000 turistas al año (cerca del 2 % del total). Eliminarla impediría analizar estas regiones y haría que la suma de los datos no coincidiera con el total oficial. Se le asigna "Otras zonas" como país y se elimina el "2" de la nota al pie.

```python
datos[ datos['pais'].isna() ]
```

```python
datos.loc[datos['region'] == 'OTRAS ZONAS2', 'pais'] = 'Otras zonas'
datos.loc[datos['region'] == 'OTRAS ZONAS2', 'region'] = 'OTRAS ZONAS'
datos.tail()
```

### Corrección de Puerto Rico

Se elimina la marca de la nota al pie en "Puerto Rico1". Según esa nota, la mayoría de los viajeros de Puerto Rico se registran como llegadas de Estados Unidos porque usan el pasaporte de ese país, lo que explica sus valores tan bajos.

Se observa un aumento inusual en 2025 (de 3 llegadas en 2024 a 156), que se revisará en el análisis exploratorio.

```python
datos.loc[datos['pais'] == 'Puerto Rico1', 'pais'] = 'Puerto Rico'
datos[datos['pais'] == 'Puerto Rico']
```

```python
datos = datos[ datos['pais'].notna() ]
datos.shape
```

### Validación

Se comprueba que la limpieza no eliminó ni duplicó datos. Para ello, se suman las llegadas de todos los países por año y se comparan con el TOTAL que reporta el ICT en el archivo original.

```python
anios = ['2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']

comparacion = pd.DataFrame({
    'suma_paises': datos[anios].sum().values,
    'total_ict': df.iloc[11, 3:12].values
}, index=anios)

comparacion['coincide'] = comparacion['suma_paises'] == comparacion['total_ict']
comparacion
```

Las sumas coinciden con el total oficial en los nueve años, por lo que no se perdieron datos durante la limpieza.

### Transformación a formato largo

La tabla tiene formato ancho: un país por fila y un año por columna. Se transforma a formato largo con `pd.melt()`, de modo que cada fila representa un país en un año, con las columnas `region`, `pais`, `anio` y `llegadas`.

Este formato (conocido como *tidy data*) facilita agrupar, filtrar y graficar los datos, ya que el año pasa a ser una variable en lugar de estar repartido en nueve columnas. El resultado tiene 495 filas (55 orígenes × 9 años).

```python
largo = pd.melt(
    datos,
    id_vars=['region', 'pais'],
    var_name='anio',
    value_name='llegadas'
)
print(largo.shape)
largo.head(10)
```

### Revisión de tipos y valores faltantes

Después de la transformación se revisa el tipo de dato de cada columna y la cantidad de valores vacíos. La columna `anio` quedó como texto y `llegadas` como `object`, por lo que ambas deben convertirse a números enteros. Además, `llegadas` tiene 4 valores vacíos que se analizan a continuación.

```python
print(largo.dtypes)
largo.isna().sum()
```

```python
largo[ largo['llegadas'].isna() ]
```

```python
datos[ datos['pais'] == 'Guyana Francesa' ]
```

### Valores faltantes

La columna `llegadas` tiene 4 valores vacíos, todos de Guyana Francesa entre 2017 y 2020. A partir de 2021 el ICT registra este país por separado (con valores entre 0 y 1), por lo que los vacíos indican que no había registro, no que no hubo llegadas. Probablemente esos viajeros se contabilizaban dentro de Francia.

Se decide conservar los valores como `NaN` para no inventar datos. La columna se convertirá al tipo `Int64`, que admite números enteros y valores vacíos. Esto no afecta el análisis, ya que las sumas y los promedios ignoran los valores vacíos.

```python
largo['anio'] = largo['anio'].astype('int')
largo['llegadas'] = largo['llegadas'].astype('Int64')
largo.dtypes
```

## 4. Guardado de datos limpios

Se guarda la tabla final en `data/processed/llegadas_por_pais.csv`. El archivo contiene 495 filas (55 orígenes × 9 años) y cuatro columnas: `region`, `pais`, `anio` y `llegadas`. Este archivo es el punto de partida para el análisis exploratorio en `02_eda.ipynb`.

```python
largo.to_csv('../data/processed/llegadas_por_pais.csv', index=False)
```

## 5. Limpieza de llegadas por vía aérea

Al revisar la hoja `Vía_aérea` se observa que tiene exactamente el mismo formato que `Todas_vías`, por lo que se aplica el mismo proceso de limpieza.

```python
df_aerea = pd.read_excel('../data/raw/Series_sitio_web_ICT_2025.xlsx', sheet_name='Vía_aérea', header=None)
```

```python
datos_aerea = df_aerea.iloc[11:78,1:12].copy()
datos_aerea.head()
```

```python
datos_aerea.columns = ['region', 'pais', '2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']
datos_aerea.head()
```

```python
datos_aerea['region'] = datos_aerea['region'].ffill()
datos_aerea.head()
```

```python
datos_aerea.loc[datos_aerea['region'] == 'OTRAS ZONAS2', 'pais'] = 'Otras zonas'
datos_aerea.loc[datos_aerea['region'] == 'OTRAS ZONAS2', 'region'] = 'OTRAS ZONAS'
datos_aerea.tail()
```

```python
datos_aerea.loc[datos_aerea['pais'] == 'Puerto Rico1', 'pais'] = 'Puerto Rico'
datos_aerea[datos_aerea['pais'] == 'Puerto Rico']
```

```python
datos_aerea = datos_aerea[ datos_aerea['pais'].notna() ]
datos_aerea.shape
```

```python
anios = ['2017', '2018', '2019', '2020', '2021', '2022', '2023', '2024', '2025']

comparacion_aerea = pd.DataFrame({
    'suma_paises': datos_aerea[anios].sum().values,
    'total_ict': df_aerea.iloc[11, 3:12].values
}, index=anios)

comparacion_aerea['coincide'] = comparacion_aerea['suma_paises'] == comparacion_aerea['total_ict']
comparacion_aerea
```

```python
largo_aerea = pd.melt(
    datos_aerea,
    id_vars=['region', 'pais'],
    var_name='anio',
    value_name='llegadas_aereas'
)
print(largo_aerea.shape)
largo_aerea.head(10)
```

```python
largo_aerea['anio'] = largo_aerea['anio'].astype('int')
largo_aerea['llegadas_aereas'] = largo_aerea['llegadas_aereas'].astype('Int64')
largo_aerea.dtypes
```

```python
largo_aerea.to_csv('../data/processed/llegadas_aereas_por_pais.csv', index=False)
```

El dataset final contiene las llegadas internacionales a Costa Rica por país de origen entre 2017 y 2025:

- **55 orígenes:** 52 países y 3 categorías agrupadas del ICT ("Otros Caribe", "Otros Europa" y "Otras zonas", que reúne Asia y el Pacífico, África y Oriente Medio).

- **Llegadas por vía aérea:** se aplicó la misma limpieza a la hoja `Vía_aérea` y se guardó en `llegadas_aereas_por_pais.csv`, con la misma estructura (495 filas) y la columna `llegadas_aereas`.
